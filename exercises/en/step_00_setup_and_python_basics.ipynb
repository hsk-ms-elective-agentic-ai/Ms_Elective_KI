## 3. The `uv` Package Manager

This project uses [`uv`](https://docs.astral.sh/uv/) — a fast, modern Python package manager — instead of plain `pip` + `venv`. It manages the virtual environment and every dependency for you from a single `pyproject.toml` file.

### Installing uv

```bash
# macOS / Linux
curl -LsSf https://astral.sh/uv/install.sh | sh

# Windows (PowerShell)
powershell -c "irm https://astral.sh/uv/install.ps1 | iex"
```

Verify it installed:

```bash
uv --version
```

### Setting up this project

From your team repo's root:

```bash
uv sync
```

This creates a `.venv/` and installs every dependency listed in `pyproject.toml` — `crewai` (agents, tasks, and the `LLM` class you'll use directly in steps 1-2), `jupyter`/`ipykernel` (these notebooks), `python-dotenv` (`.env` loading), and `google-generativeai`. Nothing to install one by one.

Then copy `.env.example` to `.env` and fill in your API keys — see "Getting your API keys" right below.

Run any command through `uv` without activating anything manually:

```bash
uv run python some_script.py
uv run jupyter notebook
```

### Register the Jupyter kernel

So VS Code/Cursor's notebook cells actually run inside this project's environment (not some other Python install), register it as a kernel once:

```bash
uv run python -m ipykernel install --user --name research_crew --display-name "research_crew"
```

Then, in any notebook (including this one), use the kernel picker in the top-right corner and select **"research_crew"**.

### Checking your environment

A couple of ways to see what's actually installed, if something feels off:

````python
import sys

print("Python executable:", sys.executable)
print("Python version:", sys.version)
````

## 4. Jupyter Notebooks

A notebook is a sequence of **cells** — each one is either **code** or **Markdown** (formatted text, like this cell). Code cells share one running Python process (the **kernel**): a variable defined in one cell is available in every cell after it — *in the order you ran them*, which is not necessarily top-to-bottom if you jump around.

**Running a cell**: click the ▷ button, or press `Shift+Enter` (run and move to the next cell) or `Ctrl+Enter` (run and stay).

**Cell shortcuts** (press `Esc` first to leave edit mode, then):

- `b` / `a` — insert a new cell below / above
- `m` / `y` — turn the current cell into Markdown / code
- `dd` — delete the current cell

Try it — run the cell below:

````python
print("Hello from the research_crew kernel!")
````

### Markdown quick reference

Markdown cells (like this one) support GitHub-flavored formatting:

